```python
import pandas as pd

# Attractions data
attractions_data = [
    ["Delhi", "Red Fort", "Heritage", 3],
    ["Delhi", "Qutub Minar", "Heritage", 2],
    ["Jaipur", "Hawa Mahal", "Heritage", 2],
    ["Jaipur", "City Palace", "Heritage", 3],
    ["Agra", "Taj Mahal", "Heritage", 4],
    ["Mumbai", "Marine Drive", "Nature", 2],
    ["Goa", "Baga Beach", "Nature", 3],
    ["Bengaluru", "Lalbagh Garden", "Nature", 2],
    ["Hyderabad", "Charminar", "Heritage", 2],
    ["Kolkata", "Victoria Memorial", "Heritage", 3]
]

attractions_df = pd.DataFrame(
    attractions_data,
    columns=["city", "attraction", "type", "avg_time_hours"]
)

# Food data
food_data = [
    ["Delhi", "North Indian", "Chole Bhature", "Yes", 150],
    ["Delhi", "Street Food", "Paratha", "Yes", 120],
    ["Jaipur", "Rajasthani", "Dal Baati Churma", "Yes", 200],
    ["Agra", "Sweet", "Petha", "Yes", 100],
    ["Mumbai", "Street Food", "Vada Pav", "Yes", 50],
    ["Goa", "Seafood", "Fish Curry", "No", 300],
    ["Bengaluru", "South Indian", "Dosa", "Yes", 120],
    ["Hyderabad", "Biryani", "Hyderabadi Biryani", "No", 250],
    ["Kolkata", "Bengali", "Rosogolla", "Yes", 120],
    ["Varanasi", "Street Food", "Kachori Sabzi", "Yes", 80]
]

food_df = pd.DataFrame(
    food_data,
    columns=["city", "food_type", "dish", "veg", "avg_price"]
)

print(" Attractions & Food data loaded")
```

```python
from google.colab import userdata

# Load WEATHERSTACK API key
WEATHERSTACK_API_KEY = userdata.get('WEATHERSTACK_API_KEY')

if WEATHERSTACK_API_KEY:
    print(" WEATHERSTACK API key retrieved!")
else:
    print(" Please add WEATHERSTACK API in Colab Secrets.")
```

```python
import requests
def get_weather(city):
    url = "http://api.weatherstack.com/current"
    params = {
        "access_key": WEATHERSTACK_API_KEY,
        "query": f"{city}, India"
    }

    response = requests.get(url, params=params)
    data = response.json()

    if "error" in data:
        print("Weatherstack error:", data["error"]["info"])
        return None

    return {
        "temperature": data["current"]["temperature"],
        "description": data["current"]["weather_descriptions"][0]
    }
```

```python
city = "Delhi"

weather = get_weather(city)

if weather is None:
    print(" Weather data not available")
else:
    print(" Weather data fetched successfully")
    print("Temperature:", weather["temperature"], "°C")
    print("Description:", weather["description"])
```

```python
import random

def get_city_attractions(city, weather_description):
    city_attractions = attractions_df[attractions_df["city"] == city]

    # If raining, avoid nature attractions
    if "rain" in weather_description.lower():
        city_attractions = city_attractions[
            city_attractions["type"] != "Nature"
        ]

    return city_attractions.sample(min(1, len(city_attractions)))

def get_city_food(city, veg_preference="Yes"):
    city_food = food_df[food_df["city"] == city]

    if veg_preference == "Yes":
        city_food = city_food[city_food["veg"] == "Yes"]

    return city_food.sample(1)
```

```python
def generate_full_itinerary(city, days, weather_description, veg_preference="Yes"):
    itinerary = []

    for day in range(1, days + 1):
        attraction = get_city_attractions(city, weather_description)
        food = get_city_food(city, veg_preference)

        if not attraction.empty:
            attraction_name = attraction.iloc[0]["attraction"]
        else:
            attraction_name = "Local markets & museums"

        if not food.empty:
            dish_name = food.iloc[0]["dish"]
        else:
            dish_name = "Local cuisine"

        day_plan = f"""
Day {day}:
Morning: Visit {attraction_name}
Lunch: Try {dish_name}
Evening: Leisure walk & shopping
Dinner: Explore local restaurants
        """

        itinerary.append(day_plan.strip())

    return itinerary
```

```python
city = "Bengaluru"
days = 1
veg_preference = "Yes"

weather = get_weather(city)

if weather is None:
    print("Weather issue, cannot plan trip")
else:
    print(" Weather:", weather["description"])
    print("\nYour Travel Itinerary:\n")

    plan = generate_full_itinerary(
        city,
        days,
        weather["description"],
        veg_preference
    )

    for day in plan:
        print(day)
        print("-" * 40)
```
